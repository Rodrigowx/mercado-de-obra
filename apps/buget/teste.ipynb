```python
unique_descriptions = analiticoFinalDf[['CODIGO DA COMPOSICAO', 'DESCRICAO DA COMPOSICAO']].drop_duplicates()

with open('unique_descriptions.txt', 'w', encoding='utf-8') as file:
    for _, row in unique_descriptions.iterrows():
        file.write(f"{row['DESCRICAO DA COMPOSICAO']} | cod: {row['CODIGO DA COMPOSICAO']}\n")
```

```python
import pandas as pd
import re

# Carregar o conteúdo
with open('unique_descriptions.txt', 'r', encoding='utf-8') as f:
    lines = [line.strip() for line in f.readlines()]

# Função de extração simplificada
def extract_details(line):
    components = {
        'Codigo': None,  # Adicionamos o código da composição
        'Material': None,
        'DN': None,
        'Junta': None,
        'Interferencia': None,
        'Fornecimento': None,
        'Detalhes': []
    }

    # Extrair o código
    match = re.search(r'\| cod: (\d+)', line)
    if match:
        components['Codigo'] = match.group(1)
        line = line.replace(match.group(0), '').strip()  # Remover o código da string principal

    # Remover AF e parênteses
    line = re.sub(r'AF_\d{2}/\d{4}|\(|\)', '', line)
    
    # Fornecimento
    if 'NÃO INCLUI FORNECIMENTO' in line:
        components['Fornecimento'] = 'Não Inclui'
        line = line.replace('NÃO INCLUI FORNECIMENTO', '')
    elif 'INCLUI FORNECIMENTO' in line:
        components['Fornecimento'] = 'Inclui'
        line = line.replace('INCLUI FORNECIMENTO', '')

    # Interferência
    if 'NÍVEL ALTO' in line:
        components['Interferencia'] = 'Alto'
        line = line.replace('NÍVEL ALTO DE INTERFERÊNCIAS', '')
    elif 'NÍVEL BAIXO' in line:
        components['Interferencia'] = 'Baixo'
        line = line.replace('NÍVEL BAIXO DE INTERFERÊNCIAS', '')

    # DN e Junta
    dn_match = re.search(r'DN[\s\dMM"]+', line)
    if dn_match:
        components['DN'] = dn_match.group().replace('DN', '').strip()
        line = line.replace(dn_match.group(), '')

    junta_match = re.search(r'JUNTA\s+\w+', line)
    if junta_match:
        components['Junta'] = junta_match.group().replace('JUNTA', '').strip()
        line = line.replace(junta_match.group(), '')

    # Material e Detalhes
    materials = ['FERRO FUNDIDO', 'AÇO CARBONO', 'PVC', 'CONCRETO', 'PEAD', 'MADEIRA']
    for material in materials:
        if material in line:
            components['Material'] = material
            line = line.replace(material, '')
            break

    components['Detalhes'] = [x.strip() for x in line.split(',') if x.strip()]
    return components

# Processar linhas
data = [extract_details(line) for line in lines]

diff = ['SAME']
for i in range(1, len(data)):
    diffs = []
    for key in data[i]:
        if key != 'Codigo':  # Ignorar a coluna Codigo na comparação
            if data[i][key] != data[i-1][key]:
                if key == 'Detalhes':
                    diffs.append(f'Detalhes: {", ".join(data[i]["Detalhes"])}')
                else:
                    diffs.append(f'{key}: {data[i][key]}')
    diff.append(' | '.join(diffs) if diffs else 'SAME')

# Criar DataFrame final
df_variacoes = pd.DataFrame(data)
df_variacoes['Diferenças'] = diff
```

```python
import pandas as pd
import json

# Função para converter listas em tuplas nas colunas especificadas
def converter_listas_em_tuplas(df):
    for coluna in df.columns:
        if df[coluna].apply(lambda x: isinstance(x, list)).any():
            df[coluna] = df[coluna].apply(lambda x: tuple(x) if isinstance(x, list) else x)
    return df

# Função para expandir composições dentro de um grupo
def expandir_composicoes_no_grupo(grupo, df_geral):
    grupo_exp = grupo.copy()
    composicoes_expandidas = set()

    while True:
        comp_items = grupo_exp[grupo_exp['TIPO ITEM'] == 'COMPOSICAO']
        if comp_items.empty:
            break

        something_expanded = False

        for _, row in comp_items.iterrows():
            codigo_item_val = row['CODIGO ITEM']
            if codigo_item_val in composicoes_expandidas:
                continue

            linhas_filhas = df_geral[df_geral['CODIGO DA COMPOSICAO'] == codigo_item_val].copy()
            if not linhas_filhas.empty:
                linhas_filhas['CODIGO DA COMPOSICAO'] = row['CODIGO DA COMPOSICAO']
                linhas_filhas['DESCRICAO DA COMPOSICAO'] = row['DESCRICAO DA COMPOSICAO']
                grupo_exp = pd.concat([grupo_exp, linhas_filhas], ignore_index=True)
                composicoes_expandidas.add(codigo_item_val)
                something_expanded = True

        # Converter listas em tuplas para evitar erros de hashabilidade
        grupo_exp = converter_listas_em_tuplas(grupo_exp)

        # Remover duplicatas considerando todas as colunas
        grupo_exp = grupo_exp.drop_duplicates()

        if not something_expanded:
            break

    return grupo_exp

# ---------------------------------------------------------------------------- #
# Execução do Processo
# ---------------------------------------------------------------------------- #

# Supondo que df_merged já esteja definido
# df_merged = ...

# Preenchendo valores nulos e criando uma cópia para trabalhar
df_filtered = df_merged.fillna('nda2').copy()

# Expandir cada grupo baseado no 'CODIGO DA COMPOSICAO'
lista_grupos = []
for cod_composicao, grupo in df_filtered.groupby('CODIGO DA COMPOSICAO'):
    grupo_exp = expandir_composicoes_no_grupo(grupo.copy(), df_filtered)
    lista_grupos.append(grupo_exp)

# Concatenar todos os grupos expandidos em um único DataFrame
df_ajustado = pd.concat(lista_grupos, ignore_index=True)

# Construir a estrutura do JSON agrupando por 'DESCRICAO DA CLASSE' e 'DESCRICAO DA COMPOSICAO'
result_dict = {}
for class_name, class_group in df_ajustado.groupby('DESCRICAO DA CLASSE'):
    result_dict[class_name] = {}
    for comp_desc, comp_group in class_group.groupby('DESCRICAO DA COMPOSICAO'):
        result_dict[class_name][comp_desc] = []
        for item_desc, item_group in comp_group.groupby('DESCRIÇÃO ITEM'):
            item_list = item_group.to_dict(orient='records')
            result_dict[class_name][comp_desc].append({
                "DESCRIÇÃO ITEM": item_desc,
                "Itens": item_list
            })

# Salvar a estrutura JSON
with open('resultado.json', 'w', encoding='utf-8') as json_file:
    json.dump(result_dict, json_file, ensure_ascii=False, indent=4)

# # Salvar o DataFrame ajustado em CSV
# df_ajustado.to_csv('df_ajustado.csv', index=False, encoding='utf-8-sig')

print("Processo concluído! JSON e DataFrame (df_ajustado) gerados.")
```
